```python
import allo
from allo.ir.types import float32

M, N, K = 32, 32, 32
def gemm(A: float32[M, K], B: float32[K, N]) -> float32[M, N]:
    C: float32[M, N] = 0.0
    for i, j in allo.grid(M, N):
        for k in allo.reduction(K):
            C[i, j] += A[i, k] * B[k, j]
    return C
```

```python
from allo.ir.utils import parse_ast

tree = parse_ast(gemm)
tree
```

```python
class Float:
    def __init__(self, bits): 
        self.bits = bits; 
        self.name = f"f{bits}"
    def __getitem__(self, shape):
        return ("float", shape)

float32 = Float(32)
```
